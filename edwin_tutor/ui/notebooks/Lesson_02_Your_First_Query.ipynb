Choose how fast/smart you want it to be:

````python
# Configuration modes

config_modes = {
    "BARE": {
        "speed": "~50ms",
        "description": "Fastest - basic processing",
        "use_case": "Quick lookups, simple queries"
    },
    "FAST": {
        "speed": "~150ms",
        "description": "Balanced - good quality + speed",
        "use_case": "Most queries (recommended)"
    },
    "FUSED": {
        "speed": "~300ms",
        "description": "Smartest - full analysis",
        "use_case": "Complex reasoning, deep analysis"
    }
}

print("⚙️ HoloLoom Configuration Modes:\n")
for mode, info in config_modes.items():
    print(f"{mode}:")
    print(f"  Speed: {info['speed']}")
    print(f"  {info['description']}")
    print(f"  Use: {info['use_case']}")
    print()

# How to use (in actual code):
print("💡 In actual code:")
print("  config = Config.bare()   # Fastest")
print("  config = Config.fast()   # Balanced (recommended)")
print("  config = Config.fused()  # Smartest")
````

## Step 5: Ask a Question (Weave a Query)

Now you can ask HoloLoom questions about what you taught it!

````python
# Example query workflow

print("🔍 Example Query Workflow:\n")
print("1. Create a query:")
print('   query = Query(text="What is Python?")\n')

print("2. Weave (process) the query:")
print('   async with WeavingOrchestrator(cfg=config, shards=shards) as shuttle:')
print('       spacetime = await shuttle.weave(query)\n')

print("3. Get the answer:")
print('   print(spacetime.response)\n')

print("📖 What happens behind the scenes:")
print("  1. HoloLoom searches its memory for relevant facts")
print("  2. It analyzes the context and relationships")
print("  3. It uses Thompson Sampling to pick the best approach")
print("  4. It synthesizes an answer from what it knows")
print("  5. It returns the result with confidence scores")
````

## Complete Example (Full Pattern)

Here's how all the pieces fit together:

````python
complete_example = '''
import asyncio
from HoloLoom.config import Config
from HoloLoom.weaving_orchestrator import WeavingOrchestrator
from HoloLoom.documentation.types import Query, MemoryShard

async def main():
    # Step 1: Teach HoloLoom some facts
    shards = [
        MemoryShard(text="Python is a programming language", source="basics"),
        MemoryShard(text="HoloLoom uses neural networks", source="tech")
    ]

    # Step 2: Configure (fast = balanced speed/quality)
    config = Config.fast()

    # Step 3: Ask a question!
    async with WeavingOrchestrator(cfg=config, shards=shards) as shuttle:
        result = await shuttle.weave(Query(text="What is Python?"))
        print(result.response)

asyncio.run(main())
'''

print("📝 Complete Example:")
print(complete_example)
print("\n💡 Try this in the main HoloLoom repository!")
````
